## 2) Listes, tranches, compréhensions et alias
- Méthodes fréquentes : `append`, `extend`, `insert`, `pop`, `remove`
- Une **tranche** `[début:fin]` **inclut** le début et **exclut** la fin
- Une **compréhension** construit une liste en une expression

```python
nums = [1, 4, 9, 16, 25]

print("tranche 1:4 ->", nums[1:4])      # indices 1, 2, 3
print("deux derniers ->", nums[-2:])

racines = [n ** 0.5 for n in nums if n % 2 == 0]   # garder les pairs
print("racines des pairs ->", racines)
```

### Attention : une affectation ne copie pas
`b = a` donne un **second nom** à la même liste. Pour une vraie copie, utilisez `a.copy()`.

```python
a = [1, 2, 3]
b = a            # alias : deux noms, une seule liste
c = a.copy()     # copie : une nouvelle liste

b.append(4)
print("a :", a, "| b :", b, "| c :", c)
print("a is b ->", a is b, "| a is c ->", a is c)
```

## 3) Tuples, dictionnaires et ensembles
- **Tuple** : `nom, age = ("Sara", 23)` (déballage)
- **Dict** : `keys()`, `values()`, `items()`, `get()`
- **Set** : opérations ensemblistes `|` (union), `&` (intersection), `-` (différence), `^`

```python
person = ("Sara", 23)
name, age = person
print(f"{name=} {age=}")

grades = {"Ali": 14.5, "Sara": 16.0, "Youssouf": 12.0}
passed = {k: (v >= 10) for k, v in grades.items()}   # compréhension de dict
print("passed :", passed)
print("Karim  :", grades.get("Karim", "absent"))     # get évite le KeyError

a = {"ai", "python", "ml"}
b = {"python", "data"}
print("union :", a | b, "| intersection :", a & b, "| différence :", a - b)
```

## 5) Lire et écrire des **CSV** (sans pandas)
- Module standard `csv` : `DictWriter` et `DictReader` travaillent **par noms de colonnes**
- On ouvre le fichier avec `newline=""`
- **Tout ce qui est lu dans un CSV est une chaîne** : il faut convertir `age` en `int` et `grade` en `float`

```python
import csv

csv_path = DATA / "students.csv"
rows = [
    {"name": "Ali", "age": 21, "grade": 14.5},
    {"name": "Sara", "age": 23, "grade": 16.0},
    {"name": "Youssouf", "age": 20, "grade": 12.0},
    {"name": "Nadia", "age": 22, "grade": 8.5},
]
with csv_path.open("w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "age", "grade"])
    writer.writeheader()
    writer.writerows(rows)

with csv_path.open(encoding="utf-8") as f:
    data = list(csv.DictReader(f))

print("avant conversion :", data[0])      # des chaînes partout
for r in data:
    r["age"] = int(r["age"])
    r["grade"] = float(r["grade"])
print("après conversion :", data[0])
```

## 6) Lire et écrire du **JSON**
- Format **structuré** : un objet JSON devient un `dict`, un tableau devient une `list`
- `json.dump` / `json.load` pour les **fichiers** ; `json.dumps` / `json.loads` pour les **chaînes**
- `indent=2` rend le fichier lisible ; `ensure_ascii=False` garde les accents

```python
import json

print(json.dumps({"ville": "Fès"}))                        # accent échappé
print(json.dumps({"ville": "Fès"}, ensure_ascii=False))    # accent conservé

json_path = DATA / "students.json"
payload = {"students": data, "meta": {"cohort": "IA-GI", "year": 2026}}

with json_path.open("w", encoding="utf-8") as f:
    json.dump(payload, f, ensure_ascii=False, indent=2)

with json_path.open(encoding="utf-8") as f:
    loaded = json.load(f)

print(loaded["meta"], "|", len(loaded["students"]), "étudiants")
```

## 7) Erreurs d'entrée‑sortie
- `FileNotFoundError` : le fichier n'existe pas
- `UnicodeDecodeError` : l'encodage choisi ne correspond pas au fichier
- Règle : **on n'attrape que l'erreur qu'on sait traiter** (jamais un `except:` nu)

```python
def safe_read_text(path, encoding="utf-8"):
    try:
        return path.read_text(encoding=encoding)
    except FileNotFoundError:
        return f"[ERREUR] Fichier introuvable : {path}"
    except UnicodeDecodeError:
        return "[ERREUR] Problème d'encodage — essayez encoding='latin-1'"


# Un fichier écrit en latin-1, lu comme s'il était en UTF-8
p_latin = DATA / "latin1.txt"
p_latin.write_text("Café à Fès", encoding="latin-1")

print(safe_read_text(DATA / "hello.txt"))
print(safe_read_text(DATA / "absent.txt"))
print(safe_read_text(p_latin))
print(safe_read_text(p_latin, encoding="latin-1"))
```
